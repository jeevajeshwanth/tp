```python
import pandas as pd
df = pd.read_csv("C:\\Users\\jeeva\\OneDrive\\Desktop\\pandas_dtset.txt")
print(df.head())
```

```python
#1
print(df.head())
print()
print(df.tail(3))
print()
print("Rows and columns:", df.shape)
print()
print(df.dtypes)
```

```python
#2
print(df[(df['Department'] == 'IT') & (df['Salary'] > 40000)])
```

```python
#3
print(df.sort_values('Salary', ascending=False)[['Name', 'Department', 'Salary']])
```

```python
#4
print("Mean salary:", df['Salary'].mean())
print("Max salary:", df['Salary'].max())
print("Min salary:", df['Salary'].min())
print("Median salary:", df['Salary'].median())
print("Avg age:", df['Age'].mean())
```

```python
#5
ans = df.groupby('Department').agg(
    EmployeeCount=('EmpID', 'count'),
    AvgSalary=('Salary', 'mean'),
    HighestSalary=('Salary', 'max')
)

print(ans.sort_values('AvgSalary', ascending=False))
```

```python
#6
df['AnnualSalary'] = df['Salary'] * 12

print(df[['Name', 'Salary', 'AnnualSalary']])
```

```python
#7
df_copy = df.copy()

df_copy.loc[0, 'City'] = None
df_copy.loc[len(df_copy)] = df_copy.iloc[1]

print(df_copy.isnull().sum())
print()
df_copy['City'] = df_copy['City'].fillna('Unknown')
df_copy = df_copy.drop_duplicates()
print(df_copy)
```

```python
#8
def salary_level(s):
    if s >= 50000:
        return 'High'
    elif s >= 40000:
        return 'Medium'
    else:
        return 'Low'

df['SalaryLevel'] = df['Salary'].apply(salary_level)
print(df)
```

```python
#9
Empmarks = pd.DataFrame({
    'EmpID': [101, 103, 105],
    'Marks': [85, 90, 78]
})

df['Marks'] = Empmarks['Marks']
print(df)
```

```python
#10
print(df.groupby('Department')['Salary'].mean().idxmax())

avg = df['Salary'].mean()
print(df[df['Salary'] > avg])
```
